```python
from pathlib import Path
import pyvista as pv

from phd_helpers.paths import (
    get_db_path, get_subject_stl_path, get_mesh, get_bone_inertia, get_relative_transform_new_basis, transform_mesh
)
```

```python
mesh_path = Path('../WorkPackages/TMCJ-Contact/Experimental/3Dprints/instronJigPipeline/outputs/CA-35T/meshes')

# small, medium, large, contact areas
subjects = ['14548R', '50000R', '50017L', '50034R']
poses = [
    'adduction','abduction','flexion','extension', 'pinch_load',
    #'pinch','grasp','jar',
    #'jar_load','grasp_load',
    'neutral'
    ]
filetype = 'stl' # for saving 
```

```python
for sub in subjects:
    savepath = Path('meshes') / sub
    savepath.mkdir(exist_ok=True, parents=True)

    pv.read(list(mesh_path.glob(f'{sub}/mc1_boneJig*.vtp'))[0]).save(savepath / f'mc1_boneJig.{filetype}')
    pv.read(list(mesh_path.glob(f'{sub}/mc1_bone_manifold*.vtp'))[0]).save(savepath / f'mc1_fullbone.{filetype}')
    pv.read(list(mesh_path.glob(f'{sub}/tpm_bone_manifold*.vtp'))[0]).save(savepath / f'tpm_fullbone.{filetype}')
    for pose in poses:
        pv.read(list(mesh_path.glob(f'{sub}/tpm_boneJig_{pose}*.vtp'))[0]).save(savepath / f'tpm_boneJig_{pose}.{filetype}')
```
