```python
from pathlib import Path
import shutil
```

```python
# get files
subs = ['50000R', '50017L', '50034R']
poses = ['pinch_load', 'flexion', 'extension', 'abduction', 'adduction']


guide_dir = Path('../guideJig/guide_meshes')

tpm_paths = sorted(list(Path('boneJigs_labelled').glob('*tpm*.3mf')))
mc1_paths = sorted(list(Path('boneJigs_labelled').glob('*mc1*.3mf')))
guide_paths = sorted(list(guide_dir.glob('*.stl')))
spacer_path = '../guideJig/spacer2mm.stl'
```

```python
# save files into root dir in sub folders
root_dir = Path('final_prints/MO_prints')
tpm_dir = root_dir / 'tpms'
mc1_dir = root_dir / 'mc1s'
guide_dir = root_dir / 'guides'


for out_dir, files in zip([tpm_dir, mc1_dir, guide_dir], [tpm_paths, mc1_paths, guide_paths+[spacer_path]]):

    out_dir.mkdir(parents=True, exist_ok=True)

    for file in files:
        shutil.copy(file, out_dir)
```

```python
# zip it
shutil.make_archive(
    base_name=str(root_dir),
    format="zip",
    root_dir=root_dir.parent,
    base_dir=root_dir.name
)
```
